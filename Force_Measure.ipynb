```python
import socket
import struct
import time
```

```python
# Function to receive data from the UR5e robot
def get_ur5e_force_data(host="172.22.22.2", port=30003):
    """
    Connect to UR5e and get force data.
    The function reads a 6D force/torque vector from the robot.
    :param host: IP address of the UR5e
    :param port: TCP port (default is 30003 for real-time data)
    :return: a list of force (Fx, Fy, Fz) and torque (Tx, Ty, Tz)
    """

    # Connect to the robot
    with socket.socket(socket.AF_INET, socket.SOCK_STREAM) as s:
        s.connect((host, port))

        # Give the robot some time to establish the connection
        time.sleep(0.1)

        # Receive data from the robot (1108 bytes typically for UR5e)
        data = s.recv(1108)

        # Parse the received data to extract force and torque information
        # The force and torque data are found starting at byte 540 and take up 48 bytes (6 floats of 8 bytes each)
        force_data = struct.unpack('!6d', data[540:588])

        # Fx, Fy, Fz, Tx, Ty, Tz
        return force_data

# Usage Example
if __name__ == "__main__":
    robot_ip = "172.22.22.2"  # Replace with your robot's IP address

    try:
        while True:
            # Get the force and torque data
            force_values = get_ur5e_force_data(host=robot_ip)

            # Print the force and torque values
            print(f"Force: Fx={force_values[0]:.2f}, Fy={force_values[1]:.2f}, Fz={force_values[2]:.2f}")
            print(f"Torque: Tx={force_values[3]:.2f}, Ty={force_values[4]:.2f}, Tz={force_values[5]:.2f}")
            
            # Add a sleep to limit the rate of data retrieval
            time.sleep(0.1)

    except KeyboardInterrupt:
        print("Exiting...")
        print("Program Stopped")
```
