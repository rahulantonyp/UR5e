```python
import socket
import time
```

```python
# Define the robot's IP address and port for URScript communication
HOST = '172.22.22.2'  # UR5e robot's IP
PORT = 30002           # Port for URScript communication (usually 30002)
```

```python
def create_socket_connection():
    """Creates a socket connection to the UR5e robot."""
    s = socket.socket(socket.AF_INET, socket.SOCK_STREAM)
    s.connect((HOST, PORT))
    return s
```

```python
def send_urscript_command(sock, command):
    """Sends a URScript command to the robot."""
    command += '\n'  # Ensure the command ends with a newline character
    sock.sendall(command.encode('utf-8'))  # Send the command as bytes
```

```python
def move_end_effector(sock, x, y, z, rx, ry, rz, acceleration=0.5, velocity=0.25):
    """
    Move the robot's end effector to a specified Cartesian position.
    
    :param sock: Socket connection to the robot.
    :param x: X coordinate (meters).
    :param y: Y coordinate (meters).
    :param z: Z coordinate (meters).
    :param rx: Rotation around X (radians).
    :param ry: Rotation around Y (radians).
    :param rz: Rotation around Z (radians).
    :param acceleration: Acceleration of the robot's motion (default: 0.5 m/s^2).
    :param velocity: Velocity of the robot's motion (default: 0.25 m/s).
    """
    # Create the URScript command to move the end effector (TCP)
    ur_script_command = f"movel(p[{x}, {y}, {z}, {rx}, {ry}, {rz}], a={acceleration}, v={velocity})"
    
    # Send the command to the robot
    send_urscript_command(sock, ur_script_command)
```

```python
def main():
    # Establish connection to the UR5e robot
    sock = create_socket_connection()

    try:
        while True:
            # Prompt the user for movement inputs
            print("Enter the movement for the end effector (in meters for position, radians for rotation):")
            x = float(input("Enter X (forward/backward): "))
            y = float(input("Enter Y (left/right): "))
            z = float(input("Enter Z (up/down): "))
            rx = float(input("Enter RX (rotation around X): "))
            ry = float(input("Enter RY (rotation around Y): "))
            rz = float(input("Enter RZ (rotation around Z): "))

            # Move the end effector to the specified target pose
            move_end_effector(sock, x, y, z, rx, ry, rz)
            time.sleep(2)  # Wait for the motion to complete

    except KeyboardInterrupt:
        print("Motion interrupted by the user.")

    except ValueError:
        print("Invalid input. Please enter numerical values.")

    finally:
        sock.close()  # Ensure the socket is closed when done
        print("Connection closed.")

if __name__ == "__main__":
    main()
```
